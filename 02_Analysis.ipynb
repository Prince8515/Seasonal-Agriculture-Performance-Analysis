## ***Crop × Season Analysis***

```python
crop_season_yield = df_clean.groupby(
    ['Crop', 'Season'],
    observed=True
)['Yield_Tonnes_Ha'].mean().unstack()

crop_season_yield.round(2)
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt='.2f'
)

plt.title('Average Yield by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.show()
```

```python
crop_season_profit = df_clean.groupby(
    ['Crop', 'Season'],
    observed=True
)['Profit_INR'].mean().unstack()

crop_season_profit.round(2)
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    crop_season_profit,
    annot=True,
    fmt='.0f'
)

plt.title('Average Profit by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df_clean,
    x='Crop',
    y='Yield_Tonnes_Ha',
    hue='Season',
    errorbar=None
)

plt.title('Average Yield by Crop Across Seasons')
plt.xlabel('Crop')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.legend(title='Season')
plt.show()
```

```python
best_crop_each_season = crop_season_yield.idxmax()

best_yield_each_season = crop_season_yield.max()

pd.DataFrame({
    'Best Crop': best_crop_each_season,
    'Average Yield': best_yield_each_season.round(2)
})
```

```python
best_profit_crop = crop_season_profit.idxmax()

best_profit_value = crop_season_profit.max()

pd.DataFrame({
    'Most Profitable Crop': best_profit_crop,
    'Average Profit': best_profit_value.round(2)
})
```

```python
# Function to identify outliers using IQR

def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound


# Check important numerical variables

for column in ['Yield_Tonnes_Ha', 'Production_Tonnes', 'Profit_INR',
               'Water_Used_m3', 'Water_Efficiency_t_per_1000m3']:

    outliers, lower, upper = find_outliers(df_clean, column)

    print(f"\n{column}")
    print("Lower Bound:", round(lower, 2))
    print("Upper Bound:", round(upper, 2))
    print("Number of Outliers:", len(outliers))
```

## Key Findings and Insights

1. **Seasonal variation**
   - Kharif recorded the highest average yield and average profit among the three seasons.
   - Rabi showed moderate yield and profitability.
   - Zaid recorded the lowest average yield and negative average profit in the dataset.

2. **Crop performance**
   - Sugarcane showed the highest average yield and highest average profit.
   - Chilli also showed strong economic performance.
   - Wheat, Rice, and Maize showed negative average profit in the dataset.
   - Crop performance varied considerably across seasons.

3. **Irrigation performance**
   - Drip irrigation recorded the highest average yield and highest average profit among the irrigation methods.
   - Rainfed farming used the least water on average and showed the highest water efficiency.
   - Flood irrigation consumed substantially more water while showing lower water efficiency.

4. **Water efficiency**
   - Water efficiency showed a strong positive relationship with yield.
   - This indicates that farms producing more output per unit of water generally recorded higher yields within this dataset.

5. **Economic performance**
   - Revenue and production were strongly associated with profit.
   - Some farms showed very high profits while others recorded substantial losses, indicating considerable variation in economic performance.

6. **Environmental factors**
   - Disease and pest risk was highest on average during Kharif.
   - Rainfall, temperature, and soil-related variables showed weaker direct correlations with yield compared with water efficiency and production-related variables.

7. **Regional variation**
   - Average yield and profit differed across states and districts.
   - Punjab showed the highest average yield among the states in the dataset.

8. **Unusual observations**
   - Several extreme observations were identified for yield, production, profit, and water efficiency.
   - These were treated as observations requiring investigation rather than automatically removing them.

### Correlation-Based Insights

The correlation analysis showed a strong positive relationship between yield and water efficiency. Yield also showed positive relationships with production and profit.

Profit had a strong positive relationship with revenue and moderate positive relationships with production, yield, and water efficiency.

However, correlation does not establish causation. In addition, some variables such as production, revenue, profit, and water efficiency are mathematically derived from other variables. Therefore, these relationships should be interpreted as associations within the dataset rather than direct causal effects.

1. **Promote efficient irrigation practices**

2. **Improve water management**
   - Water efficiency showed a strong relationship with yield.
   - Farmers and planners should focus on improving output per unit of water used.

   - Kharif showed relatively higher disease and pest risk.

7. **Use regional analysis for planning**
